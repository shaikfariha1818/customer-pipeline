```python
def process_reviews(customers):
    print("Processing text data...")

    customers['review'] = customers['review'].fillna("").str.lower()

    def sentiment(text):
        if "good" in text or "great" in text or "excellent" in text:
            return "positive"
        elif "bad" in text or "slow" in text:
            return "negative"
        else:
            return "neutral"

    customers['sentiment'] = customers['review'].apply(sentiment)

    print("Text processing done!\n")

    return customers 
```

```python

def aggregate_orders(orders):
    print("Aggregating orders...")

    spending = orders.groupby('customer')['amount'].sum().reset_index()
    spending.columns = ['name', 'total_spent']

    print("Aggregation done!\n")
    return spending
```

```python
customer_spending = aggregate_orders(orders)
```

```python
customer_spending.head()
```

```python
def merge_data(customers, spending):
    print("Merging data...")

    final_df = pd.merge(customers, spending, on='name', how='left')

    # Fill customers with no orders
    final_df['total_spent'] = final_df['total_spent'].fillna(0)

    print("Merge complete!\n")
    return final_df
```

```python
final_df = merge_data(customers, customer_spending)
```

```python
def save_data(final_df):
    final_df.to_csv("final_output.csv", index=False)
    print("Data saved as final_output.csv\n")
```

```python
final_df = merge_data(customers, customer_spending)
save_data(final_df)
```

```python

def generate_insights(final_df):
    print("Generating insights...\n")

    
    top_customer = final_df.sort_values(by='total_spent', ascending=False).iloc[0]
    print("Top Customer:")
    print(top_customer[['name', 'total_spent']])

    
    revenue_by_city = final_df.groupby('city')['total_spent'].sum()
    print("\nRevenue by City:")
    print(revenue_by_city)

    
    sentiment_count = final_df['sentiment'].value_counts()
    print("\nCustomer Sentiment:")
    print(sentiment_count)
```

```python
generate_insights(final_df)
```

```python
def run_pipeline():
    customers, orders = load_data()
    customers = clean_customers(customers)
    orders = clean_orders(orders)
    customers = process_reviews(customers)
    spending = aggregate_orders(orders)
    final_df = merge_data(customers, spending)
    save_data(final_df)
    generate_insights(final_df)
```

```python
run_pipeline()
```
